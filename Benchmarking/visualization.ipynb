```python
import pandas as pd
```

```python
criteria = ['bulkModulus','latticeConstant','defectFormationEnergy','runtime','memoryusage']
benchmarkTable = pd.DataFrame(columns=criteria)
combinations = ['GpawGpaw', 'SphinxSphinx', 'GpawSphinx', 'SphinxGpaw']
for comb in combinations:
    df = pd.read_csv('./Results/FeVac/' + comb + '/benchmarks.csv')
    df = df.tail(n=3)
    entry = df.mean().round(4)
    #entry['Workflow'] = comb
    benchmarkTable.loc[comb] = entry

reference = {'bulkModulus' : '152', 'latticeConstant' : '2.86, 2.88','defectFormationEnergy' : '0.72', 'runtime' : '-', 'memoryusage' : '-'}
benchmarkTable.loc['Reference Data'] = reference
benchmarkTable.rename(columns={'bulkModulus' : 'Bulk Modulus [GPa]', 'latticeConstant' : 'Lattice Constant [A]','defectFormationEnergy' : 'Defect Formation ENergy [eV]', 'runtime' : 'Runtime [s]', 'memoryusage' : 'Peak Memory [mb]'}, inplace=True)
benchmarkTable.head()
```
