```python
import pandas as pd
```

```python
def create_table(workflowType, reference):
    criteria = ['bulkModulus','latticeConstant','defectFormationEnergy','runtime','memoryusage']
    benchmarkTable = pd.DataFrame(columns=criteria)
    combinations = ['GpawGpaw', 'SphinxSphinx', 'GpawSphinx', 'SphinxGpaw']
    for comb in combinations:
        df = pd.read_csv('./Results/' + workflowType + '/' + comb + '/benchmarks.csv')
        df = df.tail(n=3)
        df['memoryusage'] = df['memoryusage']/1000
        entry = df.mean().round(4)
        benchmarkTable.loc[comb] = entry

    benchmarkTable.loc['Reference Data'] = reference
    benchmarkTable.rename(columns={'bulkModulus' : 'Bulk Modulus [GPa]', 'latticeConstant' : 'Lattice Constant [A]','defectFormationEnergy' : 'Defect Formation ENergy [eV]', 'runtime' : 'Runtime [s]', 'memoryusage' : 'Peak Memory [mb]'}, inplace=True)
    return benchmarkTable
```

```python
reference_FeVac = {'bulkModulus' : '152', 'latticeConstant' : '2.86, 2.88','defectFormationEnergy' : '0.72', 'runtime' : '-', 'memoryusage' : '-'}
reference_2FeVac = {'bulkModulus' : '152', 'latticeConstant' : '2.86, 2.88','defectFormationEnergy' : '1.15', 'runtime' : '-', 'memoryusage' : '-'}


table_FeVac = create_table('FeVac', reference_FeVac)
table_2FeVac = create_table('2FeVac', reference_2FeVac)
table_2FeVac2Relax = create_table('2FeVac2Relax', reference_2FeVac)
table_2FeVac2Relax.head(n=10)
```

```python
# export table
with open("table_FeVac.tex", "w", encoding="utf-8") as file:
    file.write(table_FeVac.to_latex(escape=True, float_format='%.4f'))
```
